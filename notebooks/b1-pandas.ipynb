# B1 · Tablodan satır seçmek

**Bugünün tek fikri:** bir tablodan "şu koşula uyan satırları" seçmek. 3c'de takıldığın kodların yarısı bunun üstüne kurulu.

**Kural:** her kod hücresini çalıştırmadan önce altındaki boş kutuya "bu ne yapıyor" diye tek cümle yaz. Emin değilsen yine de tahminini yaz, sonra çalıştırıp kontrol et.

## 1. Oyuncak tablo

Önce 5 satırlık, elle yazılmış küçük bir tablo. Gerçek veriyle değil bununla başlıyoruz, çünkü sonucu gözle kontrol edebiliyorsun.

`pd.DataFrame({...})`: süslü parantezin içinde her satır bir sütun: `"sütun adı": [değerler]`.

````python
import pandas as pd

oyuncak = pd.DataFrame({
    "saat":  [9, 12, 15, 18, 21],
    "gfs":   [4, 8, 12, 10, 3],
    "olcum": [6, 14, 20, 16, 5],
})
oyuncak
````

## 2. Bir sütuna bakmak

`oyuncak["olcum"]`: köşeli parantezin içine sütun adını yazınca yalnız o sütun gelir.

````python
oyuncak["olcum"]
````

## 3. Koşul sorusu sormak

Şimdi her satıra aynı soruyu soralım: "ölçüm 15 ya da daha fazla mı?" Cevap her satır için **True** (evet) ya da **False** (hayır).

**Tahmin et:** çalıştırmadan önce 5 satırın cevabını kafandan söyle (tablo yukarıda).

````python
kuvvetli = oyuncak["olcum"] >= 15
kuvvetli
````

> Bu hücre ne yapıyor? (tek cümle)
>
> oyuncak içinde sadece olcum tablosunu ve bu tablodan da 15ten büyük veya eşit true diğerlerini false döner

## 4. Cevaplarla satır seçmek

True/False listesini tablonun köşeli parantezine verince pandas yalnız **True olan satırları** bırakır. Buna **maske** deniyor: bir şablon gibi, delik olan yerden geçen görünüyor.

````python
oyuncak[kuvvetli]
````

> Hangi satırlar geldi, neden? (tek cümle)
>
> *ilk önce kuvvetli olanları bulduk daha sonra oyuncağın içine bu boolları vererek tablodaki kuvvetlilerin tüm değerlerini aldık*

## 5. Kaç tane?

`.sum()` True'ları 1, False'ları 0 sayıp toplar. Yani "kaç satır koşula uyuyor" sorusunun cevabı.

````python
kuvvetli.sum()
````

## Senin sıran: gerçek veri

Şimdi aynı şeyi 23 bin satırlık gerçek tabloda yapacaksın. Kod aynı, yalnız tablo ve sütun adları farklı. Sütunlar: `speed_kn_gfs` (GFS'in tahmini), `speed_kn_obs` (gerçek ölçüm).

````python
from ruzgar.data import load_aligned
veri = load_aligned()
veri.head()
````

**Alıştırma 1:** gerçek ölçümün 20 knot ya da üstü olduğu saatleri seç ve kaç tane olduğunu bul.
(3. ve 5. bölümdeki iki satır yeter; `oyuncak` yerine `veri`, `"olcum"` yerine `"speed_kn_obs"`.)

````python
# 1. satır: maskeyi yap
guclu_obs = veri["speed_kn_obs"]>=20
# 2. satır: kaç tane olduğunu say
guclu_obs.sum()
````

**Alıştırma 2:** **GFS'in** 20 knot ya da üstü dediği saatler kaç tane? Alıştırma 1'le karşılaştır.

````python
guclu_gfs = veri["speed_kn_gfs"]>=20
guclu_gfs.sum()
````

> İki sayı arasındaki fark sana GFS hakkında ne söylüyor? (tek cümle)
>
> *buradan anladığımız gfsnin güçlü rüzgarları(20 knot ve üstü) tahmin etmekte zorlandığı ve daha düşük tahminlerde bulundugu*

**Alıştırma 3 (bonus, sörf sorusu):** gerçekte 20+ knot esmiş, ama GFS'in 10 knot'un altında dediği saatler kaç tane? Yani GFS'in tamamen kaçırdığı kuvvetli saatler.

Yeni bir parça: iki koşulu birleştirmek için `&` ("ve") kullanılır, her koşul kendi parantezinin içinde olmalı:
```python
(kosul1) & (kosul2)
```

````python
guclu_obs = veri["speed_kn_obs"]>=20
zayif_gfs = veri["speed_kn_gfs"]<10
((guclu_obs)&(zayif_gfs)).sum()
````
